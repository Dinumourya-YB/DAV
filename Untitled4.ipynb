```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
df=pd.DataFrame({
    "Marks":[80,70,None,90,60],
    "Attendence":[90,None,85,80,75],
    "Department":["CSE","ECE","CSE","None","ECE"]
})
```

```python
#filling missing category value using mean
df["Marks"]=df["Marks"].fillna(df["Marks"].mean())
```

```python
#filling missing category value using median
df["Attendence"]=df["Attendence"].fillna(df["Attendence"].median())
```

```python
#filling missing category value using mode
df["Department"]=df["Department"].fillna(df["Department"].mode()[0])
```

```python
#label encoding
encoder=LabelEncoder()
df["Department"]=encoder.fit_transform(df["Department"])
print(df)
```
