```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("diabetes_prediction_dataset.csv")
df
```

```python
df['diabetes'].value_counts().plot(kind='line', color=['red', 'salmon'])

plt.title('Diabetes Distribution')
plt.xlabel('Diabetes (0 = No, 1 = Yes)')
plt.ylabel('Number of People')
plt.show()
```

```python
df['diabetes'].value_counts().plot(kind='bar', color=['red', 'salmon'])

plt.title('Diabetes Distribution')
plt.xlabel('Diabetes (0 = No, 1 = Yes)')
plt.ylabel('Number of People')
plt.show()
```
